```python
x = data.drop(["student_id","final_grade", "grade_category_A",
        "grade_category_A+",
        "grade_category_B",
        "grade_category_C",
        "grade_category_D",
        "grade_category_F",
        "pass_fail_Fail",
        "pass_fail_Pass"],axis=1)
y = data.final_grade
```

```python
x
```

```python
y
```

### train-test-split:

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25 , random_state=42)
```

### scaling:

```python
from sklearn.preprocessing import StandardScaler
```

```python
scalar  = StandardScaler()
```

```python
x_train_scaler = scalar.fit_transform(x_train)
```

```python
x_test_scaler  = scalar.transform(x_test)
```

```python
x_train_scaler.shape
```

```python
x_test_scaler.shape
```

### ANN(artifical neoros Networks)

```python
import tensorflow as tf
```

```python
from tensorflow import keras
```

```python
from tensorflow.keras import Sequential
```

```python
from tensorflow.keras import layers
```

```python
model_1 =  Sequential([
    layers.Input(shape=(x_train_scaler.shape[1],)),
    layers.Dense(64,activation="relu"),
    layers.Dense(32,activation="relu"),
    layers.Dense(1)
])
```

```python
model_1.summary()
```

### model_1 (optimozer = adam,activatiom=relu)

```python
model_1.compile(
    optimizer="adam",
    loss="mse",
    metrics=[
        "mae",
        tf.keras.metrics.R2Score(name="r2")
    ]
)
```

```python
model_1 .fit(x_train_scaler , y_train , epochs = 20 , batch_size=512)
```

```python
model_1.evaluate(x_test_scaler,y_test)
```

### model_2:(optimizer=sgd,activation=tanh)

```python
model_2=  Sequential([
    layers.Input(shape=(x_train_scaler.shape[1],)),
    layers.Dense(64,activation="tanh"),
    layers.Dense(32,activation="tanh"),
    layers.Dense(1)
])
```

```python
model_2.compile(
    optimizer="sgd",
    loss="mse",
    metrics=[
        "mae",
        tf.keras.metrics.R2Score(name="r2")
    ]
)
```

```python
model_2.fit(x_train_scaler,y_train,epochs=20,batch_size=256)
```

```python
model_2.evaluate(x_test_scaler,y_test)
```
